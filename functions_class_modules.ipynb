positional argument

```python
def add_positional(a,b,c):
    sum = a+b+c
    return sum
result = add_positional(2,4,6)
result
```

keyword Argument

```python
def add_keyword(name = 'alice', age='40'):
    print('name', name)
    print('age', age)

add_keyword(name = 'femi', age = 30)
```

args argumments


```python
def add_args(*args):
    print(type(args))
    for num in args:
        print(num, end = " ")
```

```python
add_args(1,2,3,4,3,5,6,7)
```

### **kwargs arguments

```python
def add_kwargs(**kwargs):
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
add_kwargs(name="femi", age = 35, city = 'ibadan')
```

```python
def average(number):
    total = 0
    for a in number:
        total += a
    avg = total/len(number)
    return avg 
```

```python
price = [12, 5 , 23, 17, 65, 82]
avg = average(price)
print (avg)
```

```python
# get cryptoprice change percentage
def calculate_price_changes(old_price, new_price):
    change = ((new_price - old_price)/old_price) * 100
    return change
```

```python
btc_change = calculate_price_changes(100000, 120000)
print(f'btc price change {btc_change} %')
```

```python
def format_crypto_price(name, price, symbol = '$'):
    return f'{name} is currently trading at {symbol}{price:,.2f}'

print(format_crypto_price('bitcoin', 117000))
print(format_crypto_price('ethereum', 3250))
```
